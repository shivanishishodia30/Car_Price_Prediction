```python
# loading libraries
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler # to handle the categorical data 
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
```

```python
# data loading 
df=pd.read_csv(r"C:\Users\INDIAN\Downloads\car_price_prediction\cardekho.csv")
df.head()
```

```python
# rename mileage(km/ltr/kg) to mileage

df.rename(columns={"mileage(km/ltr/kg)":"mileage"},inplace=True)
```

```python
df.head()
```

```python
# chk for the null values
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum(),df.shape
```

```python
# handling categorical data
# fuel	seller_type	transmission	owner
label=LabelEncoder()  # initialize label encoder 

df["fuel_encoded"]=label.fit_transform(df["fuel"])
var=dict(zip(df["fuel"],df["fuel_encoded"]))
var       
```

```python
df["seller_type_encoded"]=label.fit_transform(df["seller_type"])
var1=dict(zip(df["seller_type"],df["seller_type_encoded"]))
var1
```

```python
df["transmission_encoded"]=label.fit_transform(df["transmission"])
var2=dict(zip(df["transmission"],df["transmission_encoded"]))
var2
```

```python
df["owner_encoded"]=label.fit_transform(df["owner"])
var3=dict(zip(df["owner"],df["owner_encoded"]))
var3
```

```python
df.head()
```

```python
# dropping orignal columns

df.drop(columns=["name","fuel","seller_type","transmission","owner"],inplace=True)
```

```python
df.head()
```

```python
#feature selection
x=df.drop(columns=["selling_price","max_power"])
y=df["selling_price"]
x.head()
```

```python
# data splitting 

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection 
model=LinearRegression()
model2=RandomForestRegressor(max_depth=5,n_estimators=200,random_state=42)

# model fit 
model2.fit(x_train,y_train)
```

```python
# model testing 
y_pred=model2.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# traing accuracy 
y_pred=model2.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# data scaling 
scalar=StandardScaler()

x_train_scaled=scalar.fit_transform(x_train)
x_test_scaled=scalar.transform(x_test)  # will do transfrom on test data to avoid data leakeage 
```

```python
# model traing on scaled data 
model2.fit(x_train_scaled,y_train)
```

```python
# model testing 
y_pred=model2.predict(x_test_scaled)
r2_score(y_test,y_pred)
```

```python
# traing data 
y_pred=model2.predict(x_train_scaled)
r2_score(y_train,y_pred)
```
